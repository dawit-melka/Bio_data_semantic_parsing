```python

class PubMedDataExtractor:
    def __init__(self, data_dir, geo_output_file, sample_output_file):
        self.data_dir = data_dir
        self.geo_output_file = geo_output_file
        self.sample_output_file = sample_output_file

    def extract_geo_data(self, file_path):
        """Extracts PubMed ID and GEO Accession from a .soft.gz file."""
        pubmed_id = None
        geo_accession = None
        
        try:
            with gzip.open(file_path, "rt", encoding="utf-8") as f:
                for line in f:
                    if line.startswith("^SERIES = GSE"):
                        geo_accession = line.strip().split(" = ")[1]
                    elif line.startswith("!Series_pubmed_id ="):
                        pubmed_id = line.strip().split(" = ")[1]
                    
                    if pubmed_id and geo_accession:
                        break
        except Exception as e:
            print(f"Error processing {file_path}: {e}")

        return pubmed_id if pubmed_id else "null", geo_accession

    def extract_sample_data(self, file_path):
        """Extracts PubMed ID and Reference Series from a .soft.gz file."""
        pubmed_id = None
        reference_series = None
        
        try:
            with gzip.open(file_path, "rt", encoding="utf-8") as f:
                for line in f:
                    line = line.strip()
                    if line.startswith("!dataset_reference_series = GSE"):
                        reference_series = line.split(" = ")[-1]
                    elif line.startswith("!dataset_pubmed_id ="):
                        pubmed_id = line.split(" = ")[-1]
                    if pubmed_id and reference_series:
                        break
        except Exception as e:
            print(f"Error processing {file_path}: {e}")

        return pubmed_id if pubmed_id else "null", reference_series if reference_series else "null"

    def process_geo_directory(self):
        """Processes .soft.gz files for GEO Accession data and saves to CSV."""
        with open(self.geo_output_file, mode="w", newline="") as file:
            writer = csv.writer(file)
            writer.writerow(["File Name", "GEO Accession", "PubMed ID"])

            for filename in os.listdir(self.data_dir):
                if filename.endswith("_family.soft.gz"):
                    file_path = os.path.join(self.data_dir, filename)
                    pubmed_id, geo_accession = self.extract_geo_data(file_path)
                    
                    if pubmed_id or geo_accession:
                        writer.writerow([filename, geo_accession, pubmed_id])
                        print(f"Processed {filename}: GEO Accession: {geo_accession}, PubMed ID: {pubmed_id}")
        
        print(f"\nData saved to {self.geo_output_file}")

    def process_sample_directory(self):
        """Processes .soft.gz files for Sample data and saves to CSV."""
        with open(self.sample_output_file, mode="w", newline="") as file:
            writer = csv.writer(file)
            writer.writerow(["File Name", "Reference Series", "PubMed ID"])

            for filename in os.listdir(self.data_dir):
                if filename.startswith("GDS"):
                    file_path = os.path.join(self.data_dir, filename)
                    pubmed_id, reference_series = self.extract_sample_data(file_path)
                    if pubmed_id != "null" or reference_series != "null":
                        writer.writerow([filename, reference_series, pubmed_id])
                        print(f"Processed: {filename}: Reference Series: {reference_series}, PubMed ID: {pubmed_id}")
        
        print(f"\nData saved to {self.sample_output_file}")

if __name__ == "__main__":
    DATA_DIR = "../data_gene/"
    OUTPUT_FILE_GEO = "../data/pubmed_geo_data.csv"
    OUTPUT_FILE_SAMPLE = "../data/pubmed_sample_data.csv"
    extractor = PubMedDataExtractor(DATA_DIR, OUTPUT_FILE_GEO, OUTPUT_FILE_SAMPLE)
    extractor.process_geo_directory()
 
```
